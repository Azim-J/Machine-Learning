```python
'''
Features application:
Crime rate: Higher crime rate = lower house value
Residential Zone: Less offices nearby = lower house value
Industrial Areas: More offices nearby = higher house value
Charles river indicator: river can mean easier transport and possibly more offices
NOX: higher Nox = lower house value
RM: More rooms = higher house value
DIS: Smaller distance = higher house value
RAD: Higher highway accessibility = higher house value
TAX: More tax = higher house value 
PTRATIO: lower ratios
B: Possible racial bias at the time
LSTAT: Lower economic state = Lower house value
'''
sns.boxplot(data=df[["CRIM", "ZN"]])
```
